```python
!pip install pyspark

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.ml.classification import LogisticRegression, RandomForestClassifier
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator

spark = SparkSession.builder \
    .appName("Telco_Churn_Spark_MLlib") \
    .getOrCreate()

print("Spark Session Berhasil Dibuat!")
```

```python
!wget -O telco_churn.csv https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv

df = spark.read.csv("telco_churn.csv", header=True, inferSchema=True)
df.printSchema()
df.show(5)
```

```python
# Blok 3: Praproses Data & Feature Engineering (Perbaikan)
from pyspark.sql.functions import col, when, trim
from pyspark.ml.feature import StringIndexer, VectorAssembler

# Hapus kolom ID
df_clean = df.drop("customerID")

# Tangani string kosong/spasi pada TotalCharges: ubah spasi jadi NULL lalu cast ke double
df_clean = df_clean.withColumn(
    "TotalCharges",
    when(trim(col("TotalCharges")) == "", None)
    .otherwise(col("TotalCharges"))
    .cast("double")
)

# Hapus baris yang memiliki nilai NULL
df_clean = df_clean.na.drop()

# Konversi label Churn (Yes -> 1.0, No -> 0.0)
df_clean = df_clean.withColumn("label", when(col("Churn") == "Yes", 1.0).otherwise(0.0))

# Indexing kolom kategorikal
cat_cols = ["gender", "Partner", "Dependents", "PhoneService", "MultipleLines",
            "InternetService", "OnlineSecurity", "OnlineBackup", "DeviceProtection",
            "TechSupport", "StreamingTV", "StreamingMovies", "Contract",
            "PaperlessBilling", "PaymentMethod"]

indexed_cols = [f"{c}_idx" for c in cat_cols]
for c, idx_c in zip(cat_cols, indexed_cols):
    indexer = StringIndexer(inputCol=c, outputCol=idx_c)
    df_clean = indexer.fit(df_clean).transform(df_clean)

# Gabungkan fitur ke VectorAssembler
num_cols = ["SeniorCitizen", "tenure", "MonthlyCharges", "TotalCharges"]
feature_cols = num_cols + indexed_cols

assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
final_df = assembler.transform(df_clean).select("features", "label")

# Split dataset
train_data, test_data = final_df.randomSplit([0.8, 0.2], seed=42)
print(f"Data Latih: {train_data.count()}, Data Uji: {test_data.count()}")
```

```python
# 1. Logistic Regression
lr = LogisticRegression(featuresCol="features", labelCol="label", maxIter=10)
lr_model = lr.fit(train_data)
lr_preds = lr_model.transform(test_data)

# 2. Random Forest Classifier
rf = RandomForestClassifier(featuresCol="features", labelCol="label", numTrees=50, seed=42)
rf_model = rf.fit(train_data)
rf_preds = rf_model.transform(test_data)

# Evaluator
eval_auc = BinaryClassificationEvaluator(labelCol="label", rawPredictionCol="rawPrediction", metricName="areaUnderROC")
eval_acc = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="accuracy")
eval_f1 = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="f1")

print("HASIL LOGISTIC REGRESSION:")
print(f"Accuracy : {eval_acc.evaluate(lr_preds):.4f}")
print(f"AUC ROC  : {eval_auc.evaluate(lr_preds):.4f}")
print(f"F1-Score : {eval_f1.evaluate(lr_preds):.4f}")

print("\nHASIL RANDOM FOREST:")
print(f"Accuracy : {eval_acc.evaluate(rf_preds):.4f}")
print(f"AUC ROC  : {eval_auc.evaluate(rf_preds):.4f}")
print(f"F1-Score : {eval_f1.evaluate(rf_preds):.4f}")
```
